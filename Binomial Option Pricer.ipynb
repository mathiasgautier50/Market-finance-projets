```python
import numpy as np

S0 = 100
r = 0.02
b = 0.005
q = 0.005
sigma = 0.1
T = 1
n = 1
K=100

def pricer(option_type = "C",exercise_type = "E",S0=100,K=100,T=1,n=252,sigma=0,r=0,b=0,q=0):

    delta_t = 1/n
    u = np.exp(sigma*np.sqrt(delta_t))
    d = 1/u

    S = np.zeros((n*T+1,n*T+1))

    # Arbre des prix du ss-jacent
    for i in range(n*T+1):
        for j in range(i+1):
            S[i,j] = S0*(d**(j))*(u**(i-j))

    # Matrice avec valeurs intrinsèques à chaque instant
    O_intrinsic = np.zeros((n*T+1,n*T+1))
    if option_type == "C":
        O_intrinsic[:,:] = np.maximum(S[:,:]-K,0)
    elif option_type == "P":
        O_intrinsic[:,:] = np.maximum(K-S[:,:],0)

    # Initialisation de l'arbre
    O = np.zeros((n*T+1,n*T+1))
    O[n*T,:] = O_intrinsic[n*T,:]

    # Remontée de l'arbre
    assert delta_t<(sigma**2)/((r-b-q)**2),"la probabilité n'est pas comprise entre 0 et 1"
    p = (np.exp((r-q-b)*delta_t)-d)/(u-d)

    #Option européenne
    if exercise_type == "E":
        for i in range(n*T):
            O[n*T-1-i,:n*T-i] = np.exp(-r*delta_t)*(p*O[n*T-i,:n*T-i]+(1-p)*O[n*T-i,1:n*T-i+1])

    #Option américaine
    elif exercise_type == "A":
        for i in range(n*T):
            O[n*T-1-i,:n*T-i] = np.maximum(np.exp(-r*delta_t)*(p*O[n*T-i,:n*T-i]+(1-p)*O[n*T-i,1:n*T-i+1]),O_intrinsic[n*T-1-i,:n*T-i])
    
    return O[0,0], O
```

```python
prix,matrice = pricer(S0=100,K=80,T=3,n=252,sigma=0.15,r=0.025,b=0.0037,q=0.005)

print(prix)
print(matrice)
```
